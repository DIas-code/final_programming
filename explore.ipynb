## Осмотр датасета

```python
import pandas as pd
from src.config import CSV_PATH, ENCODING

#для осмотра только 1000 строк, все satatypes str, pd не будет сам гадать
df = pd.read_csv(CSV_PATH, nrows=1000, encoding=ENCODING, dtype=str)

# Длинный текст в ячейке не обрезается до 50 символов
pd.set_option("display.max_colwidth", None)
```

```python
print("Размер выборки (строки, колонки):", df.shape)
```

```python
print("Colums")
for col in df.columns:
    print(col)
```

```python
# T (транспонирование) переворачиваем колнки и строки
print("Первая строка")
df.head(2).T
```

```python
# проверяем даты и их Nan
date_cols = ["date_adopted", "date_modified", "date_published", "ministry_registration_date"]
df[date_cols].head(10)
```

```python
for col in ["is_repealed", "is_archive"]:
    print(df[col].value_counts(dropna=False), "\n")
```

```python
# проверяем, как разделены сферы в legal_relation_area
for value in df["legal_relation_area"].iloc[:5]:
    print(value, "\n")
```

```python
# проверяем, что пишут в скобках в adopting_body
for value in df[df["adopting_body"].str.contains(r"\(", na=False)]["adopting_body"].head(5):
    print(value, "\n")
```

**Результат:**
- `legal_relation_area`: в одной ячейке несколько сфер. Новая сфера начинается после запятой с заглавной буквы. Запятая перед маленькой буквой — это часть названия, по ней делить нельзя.
- `adopting_body`: в скобках справка о том, когда орган создан и как назывался раньше. Скобки надо убрать, иначе один орган посчитается как несколько разных. Иногда после скобки через запятую идёт второй орган — значит, акт принят двумя органами вместе.

```python
# проверяем, какие бывают формы акта
df["act_form"].value_counts().head(15)
```

```python
# проверяем, сколько разных значений в главных колонках
cols = ["act_form", "legal_force", "region", "adopting_body", "legal_relation_area"]
df[cols].nunique()
```

## Итог

- **Даты** в формате `ГГГГ-ММ-ДД`, без времени. `date_adopted` заполнена, в остальных датах много пропусков.
- **`is_repealed`** — строки `t`/`f`, их нужно превратить в `True`/`False`. **`is_archive`** везде `f`, колонку можно убрать.
- **Заглушки** `ЖОҚ` и `undefined` — это пропуски, записанные текстом.
- **Неатомарные поля:** `legal_relation_area`, `act_form`, `adopting_body` — в одной ячейке несколько значений через запятую. Их надо делить и выносить в отдельные таблицы.
- **Атомарные поля:** `legal_force`, `region` — станут обычными справочниками.
- Выборка — первые 1000 строк, она не отражает весь файл (например, отменённых актов тут 4 %, а во всём файле 35 %). Полный аудит делаем по всем строкам.
